```python
import pandas as pd
import yfinance as yf
pd.options.display.float_format = "{:,.0f}".format  # no scientific notation

tickers = ["TTWO", "UBI.PA", "9697.T"]

for t in tickers:
    company = yf.Ticker(t)
    income = company.income_stmt       # annual income statement
    balance = company.balance_sheet
    cashflow = company.cashflow
    print(t, company.info.get("currency"))
    print(income.head())
```

```python
income_items = ["Total Revenue", "Gross Profit", "Operating Income", "Net Income", "EBITDA"]

frames = []
for t in tickers:
    company = yf.Ticker(t)
    df = company.income_stmt.reindex(income_items).T   # years become rows
    df = df.dropna(how="all")                          # drop empty years
    df["ticker"] = t
    df["currency"] = company.info.get("currency")
    frames.append(df)

income = pd.concat(frames).reset_index(names="fiscal_year_end")
income
```

```python
income.columns = (income.columns.str.lower()
                                .str.replace(" ", "_"))
income["fiscal_year_end"] = pd.to_datetime(income["fiscal_year_end"]).dt.date
income
```

```python
balance_items = ["Total Assets", "Current Assets", "Current Liabilities",
                 "Total Liabilities Net Minority Interest", "Stockholders Equity",
                 "Cash And Cash Equivalents", "Total Debt", "Net Debt",
                 "Accounts Receivable", "Working Capital",
                 "Goodwill And Other Intangible Assets"]

frames = []
for t in tickers:
    company = yf.Ticker(t)
    df = company.balance_sheet.reindex(balance_items).T   # years become rows
    df = df.dropna(how="all")                             # drop empty years
    df["ticker"] = t
    df["currency"] = company.info.get("currency")
    frames.append(df)

balance = pd.concat(frames).reset_index(names="fiscal_year_end")
balance["fiscal_year_end"] = pd.to_datetime(balance["fiscal_year_end"]).dt.date
balance.columns = balance.columns.str.lower().str.replace(" ", "_")
balance = balance.dropna(subset=["total_assets"]).reset_index(drop=True)
balance["net_debt"] = balance["total_debt"].fillna(0) - balance["cash_and_cash_equivalents"]
balance
```

```python
cashflow_items = ["Operating Cash Flow", "Investing Cash Flow", "Financing Cash Flow",
                  "Capital Expenditure", "Free Cash Flow",
                  "Depreciation And Amortization",
                  "Repurchase Of Capital Stock", "Cash Dividends Paid"]

frames = []
for t in tickers:
    company = yf.Ticker(t)
    df = company.cashflow.reindex(cashflow_items).T       # years become rows
    df = df.dropna(how="all")                             # drop empty years
    df["ticker"] = t
    df["currency"] = company.info.get("currency")
    frames.append(df)

cashflow = pd.concat(frames).reset_index(names="fiscal_year_end")
cashflow["fiscal_year_end"] = pd.to_datetime(cashflow["fiscal_year_end"]).dt.date
cashflow.columns = cashflow.columns.str.lower().str.replace(" ", "_")
cashflow = cashflow.dropna(subset=["operating_cash_flow"])
cashflow[["repurchase_of_capital_stock", "cash_dividends_paid"]] = (cashflow[["repurchase_of_capital_stock", "cash_dividends_paid"]].fillna(0))
cashflow = cashflow.reset_index(drop=True)
cashflow
```

```python
# Create the companies table

companies = pd.DataFrame({
    "ticker": ["TTWO", "UBI.PA", "9697.T"],
    "company_name": ["Take-Two Interactive", "Ubisoft", "Capcom"],
    "country": ["USA", "France", "Japan"],
    "currency": ["USD", "EUR", "JPY"],
    "fiscal_year_end_month": ["March", "March", "March"]
})
companies
```

```python
for df in [income, balance, cashflow]:
    df["fiscal_year"] = pd.to_datetime(df["fiscal_year_end"]).dt.year
```

```python
for name, df in [("income", income), ("balance", balance), ("cashflow", cashflow)]:
    print(name, df.duplicated(subset=["ticker", "fiscal_year_end"]).sum())
```

```python
companies.to_csv("companies.csv", index=False)
income.to_csv("income_statement.csv", index=False)
balance.to_csv("balance_sheet.csv", index=False)
cashflow.to_csv("cash_flow.csv", index=False)
```

```python
fx_tickers = {"EUR": "EURUSD=X", "JPY": "JPYUSD=X"}

fx = yf.download(list(fx_tickers.values()), start="2022-01-01",
                 end="2026-04-01", progress=False)["Close"]
fx = fx.rename(columns={v: k for k, v in fx_tickers.items()})
fx["USD"] = 1.0          
fx.tail()
```

```python
periods = income[["currency", "fiscal_year_end"]].drop_duplicates()

rows = []
for _, r in periods.iterrows():
    end = pd.Timestamp(r["fiscal_year_end"])
    start = end - pd.DateOffset(years=1) + pd.Timedelta(days=1)
    rates = fx[r["currency"]].dropna()
    rows.append({
        "currency": r["currency"],
        "fiscal_year_end": r["fiscal_year_end"],
        "avg_rate": rates.loc[start:end].mean(), # for income and cash flow
        "closing_rate": rates.loc[:end].iloc[-1], # for balance sheet
    })

fx_rates = pd.DataFrame(rows)
pd.options.display.float_format = "{:,.6f}".format
fx_rates
```
